```python
import pandas as pd
import matplotlib.pyplot as plt

dict_faturamento = {
    'data_ref': [
        '2023-01-01',
        '2020-02-01',
        '2021-03-01',
        '2022-04-01',
        '2023-05-01',
        '2023-06-01',
        '2020-07-01',
        '2021-08-01',
        '2022-09-01',
        '2023-10-01',
        '2022-11-01',
        '2023-12-01',
    ],
    'valor': [
        400000,
        890000,
        760000,
        430000,
        920000,
        340000,
        800000,
        500000,
        200000,
        900000,
        570000,
        995000,
    ]
}
```

```python
df_faturamento = pd.DataFrame(dict_faturamento)
```

```python
df_faturamento['data_ref'] = pd.to_datetime(df_faturamento['data_ref'])
```

```python
media_vendas = df_faturamento['valor'].mean()
print(f"Média de vendas: R$ {media_vendas:,.2f}")
```

```python
df_faturamento = df_faturamento.sort_values('data_ref')
```

```python
plt.figure(figsize=(12, 6))

plt.bar(
    df_faturamento['data_ref'].dt.strftime('%m/%Y'),
    df_faturamento['valor']
)

plt.title('Faturamento por mês')
plt.xlabel('Mês de referência')
plt.ylabel('Valor (R$)')

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 6))

plt.plot(
    df_faturamento['data_ref'].dt.strftime('%m/%Y'),
    df_faturamento['valor'],
    marker='o'
)

plt.title('Evolução do faturamento')
plt.xlabel('Mês de referência')
plt.ylabel('Valor (R$)')

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```
